```python
# Question 39: Find the most common word in "the quick brown fox jumps over the lazy dog"
s = "the quick brown fox jumps over the lazy dog"

words = s.split()

max_word = ""
max_count = 0

for word in words:
    count = 0
    
    for w in words:
        if w == word:
            count += 1
    
    if count > max_count:
        max_count = count
        max_word = word

print(max_word)
```

```python
s = "((()))"

count = 0
balanced = True

for ch in s:
    if ch == "(":
        count += 1
    elif ch == ")":
        count -= 1
    
    if count < 0:
        balanced = False
        break

if count != 0:
    balanced = False

if balanced:
    print("Balanced")
else:
    print("Not Balanced")
```

```python
# Question 42: Convert "hello world" to Morse code
morse = {
    'a': '.-', 'b': '-...', 'c': '-.-.', 'd': '-..',
    'e': '.', 'f': '..-.', 'g': '--.', 'h': '....',
    'i': '..', 'j': '.---', 'k': '-.-', 'l': '.-..',
    'm': '--', 'n': '-.', 'o': '---', 'p': '.--.',
    'q': '--.-', 'r': '.-.', 's': '...', 't': '-',
    'u': '..-', 'v': '...-', 'w': '.--', 'x': '-..-',
    'y': '-.--', 'z': '--..',
    ' ': '/'
}

s = "hello world"

result = ""

for ch in s:
    result += morse[ch] + " "

print(result)
```

```python
# Question 43: Find the longest common substring between "programming" and "grammar"
s1 = "programming"
s2 = "grammar"

longest = ""

for i in range(len(s1)):
    for j in range(i, len(s1)):
        substr = s1[i:j+1]
        
        if substr in s2:
            if len(substr) > len(longest):
                longest = substr

print(longest)
```

```python
# Question 46: Convert "hello world" to Pig Latin
#pig latin means - The program converts each word by moving
# the first letter to the end and adding "ay", 
# then joins all words back into a sentence.
s = "hello world"

words = s.split()

result = []

for word in words:
    pig = word[1:] + word[0] + "ay"
    result.append(pig)

print(" ".join(result))
```

```python
## Question 47: Check if string is a valid IPv4 address: "192.168.1.1"

ip = "192.168.1.1"

parts = ip.split(".")

valid = True

if len(parts) != 4:
    valid = False
else:
    for part in parts:
        if not part.isdigit():
            valid = False
            break
        num = int(part)
        if num < 0 or num > 255:
            valid = False
            break

if valid:
    print("Valid IPv4")
else:
    print("Invalid IPv4")
```

```python
# Question 49: Convert "hello world" to ROT13 encoding

s = "hello world"

result = ""

for ch in s:
    if ch.isalpha():
        start = ord('a') if ch.islower() else ord('A')
        result += chr((ord(ch) - start + 13) % 26 + start)
    else:
        result += ch

print(result)
```

```python
# Question 50: Check if string is a valid credit card number: "4532015112830366"

card = "4532015112830366"

def is_valid(card):
    total = 0
    reverse = card[::-1]

    for i in range(len(reverse)):
        digit = int(reverse[i])

        if i % 2 == 1:
            digit *= 2
            if digit > 9:
                digit -= 9

        total += digit

    return total % 10 == 0

if is_valid(card):
    print("Valid card")
else:
    print("Invalid card")
```
