```python
# Question 1: Create a set of first 10 natural numbers
print("Question 1: Create a set of first 10 natural numbers")
s1 = set(range(1, 11))
print(s1)
```

```python
# Question 2: Add element 11 to set {1, 2, 3, 4, 5}
print("\nQuestion 2: Add element 11 to set {1, 2, 3, 4, 5}")
s2 = {1, 2, 3, 4, 5}
s2.add(11)
print(s2)
```

```python
# Question 3: Remove element 3 from set {1, 2, 3, 4, 5, 6}
print("\nQuestion 3: Remove element 3 from set {1, 2, 3, 4, 5, 6}")
s3 = {1, 2, 3, 4, 5, 6}
s3.remove(3)
print(s3)
```

```python
# Question 4: Find the intersection
print("\nQuestion 4: Find the intersection")
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}
print(a & b)
```

```python
# Question 5: Find the difference
print("\nQuestion 5: Find the difference")
print(a - b)
```

```python
# Question 6: Check if 5 is in set
print("\nQuestion 6: Check if 5 is in set")
s6 = {1, 2, 3, 4, 6, 7, 8}
print(5 in s6)
```

```python
# Question 7: Find length of set
print("\nQuestion 7: Find the length of set")
s7 = {'a', 'b', 'c', 'd', 'e'}
print(len(s7))
```

```python
# Question 8: Create set of vowels from string
print("\nQuestion 8: Create a set of vowels")
text = "hello world"
vowels = {'a', 'e', 'i', 'o', 'u'}
result = set([ch for ch in text if ch in vowels])
print(result)
```

```python
# Question 9: Remove duplicates from list
print("\nQuestion 9: Remove duplicates")
lst = [1, 2, 2, 3, 4, 4, 5, 6, 6, 7]
print(set(lst))
```

```python
# Question 10: Check subset
print("\nQuestion 10: Check subset")
s10a = {1, 2, 3}
s10b = {1, 2, 3, 4, 5, 6}
print(s10a.issubset(s10b))
```
