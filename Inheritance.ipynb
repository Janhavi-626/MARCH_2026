## Inhertance..


Definition: Inheritance is a mechanism where a new class (child/derived class) can inherit properties and methods from
an existing class (parent/base class). This promotes code reusability and establishes a relationship between classes.

```python
## Basic syntax

## Parent/ Base/ Super class..

class ParentClass:
  def __init__(self):
    pass

  # methods specific to parent class
  def parent_method(self):
    pass


## Child/ Derived/ Sub class
class ChildClass(ParentClass):
  def __init__(self):
    super().__init__()
    pass

  def child_method(self):
    # child-specific methods
    pass
```

Key Points:


Child class name is followed by (ParentClass) in definition
super() is used to access parent class methods

Child inherits ALL public and protected members from parent

Child can add its own attributes and methods

Child can override parent methods

```python
#Inheritance example
class Animal:
    def __init__(self,name,age):
        self.name=name
        self.age=age
        #methods 1.eat 2.sleep 3.make_sound
        #above all are parent class methods
    def eat(self):
            print(f"{self.name} is eating")
    def sleep(self):
            print(f"{self.name}is sleeping")
    def make_sound(self):
            print(f"{self.name}is making sound ")
```

```python
#child class
class Dog(Animal):
    def __init__(self,name,age,breed):
        super().__init__(name,age)
        self.breed=breed
#child class methods
#override parent methods
    def make_sound(self):
      print(f"{self.name}is Barking")
#new method for child class is fetch
    def fetch(self):
        print(f"{self.name}is fetching for ball")
```

```python
#2 nd child class
class Cat(Animal):
    def __init__(self,name,age,color):
        super().__init__(name,age)
        self.color=color
    #Override the parent method
    def make_sound(self):
        print(f"{self.name}is a cat it makes sound as meow")

    #2nd child specific method to cat
    def scratch(self):
        print(f"{self.name}is Scratching")
```

## Super()


## Understanding `super()`

### What is `super()`?
`super()` is a built-in function in Python that returns a temporary object of the parent class, allowing you to call its methods.

---

### Why Use `super()`?

1. **Avoid hardcoding the parent class name**  
   Makes your code more flexible and maintainable.

2. **Properly initialize parent class attributes**  
   Ensures correct initialization when working with inheritance.

3. **Call parent methods that you're overriding**  
   Useful when extending or modifying existing functionality.

4. **Essential for multiple inheritance**  
   Helps maintain the correct method resolution order (MRO).

## Multi-level inheritance..

Multi-level inheritance is when a class is derived from another derived class.

# single level inheritance.

parent -> child

# Multi-level inhertance.

Grandparent -> Parent -> Child



The Child class gets:

Its own methods

Parent class methods

Grandparent class methods

This creates a chain of inheritance

```python
#example for multi level inhritence
class Vehicle:
    def __init__(self,brand):
        self.brand=brand
    def start(self):
        print(f"{self.brand}vehicle started")
```

```python
class Car(Vehicle):
    def __init__(self,brand,model):
        super().__init__(brand)
        self.model=model

    def drive(self):
        print(f"{self.brand} {self.model}is driving")
```

```python
class ElectricCar(Car):
    def __init__(self,brand,model,battery_capacity):
        super().__init__(brand,model)
        self.battery_capacity=battery_capacity
        #electric car specific method
    def charge(self):
        print(f"{self.brand}{self.model}is charging({self.battery_capacity}kwh)")
```

```python
electriccar=ElectricCar("Telsa","Model 3",87)
```

```python
electriccar.start()
```

```python
electriccar.drive()
```

```python
electriccar.charge()
```

### Initialization Flow in Multi-Level Inheritance

Execution happens like this:

1. `ElectricCar.__init__()`
2. → calls `Car.__init__()` using `super()`
3. → calls `Vehicle.__init__()` using `super()`

---

**Conclusion:**

Initialization flows from **top to bottom** through the inheritance chain using `super()`.

## Multiple Inheritance..

Multiple inheritance is when a class inherits from more than one parent class.




```python
# Syantax

class Parent1:
  pass


class Parent2:
  pass

## Inheriting both parents at once..

class Child(Parent1, Parent2):
  pass
```

### Characteristics of Multiple Inheritance

- Child inherits from multiple parent classes  
- Can access methods and attributes from all parent classes  
- Must handle method conflicts carefully (same method name in parents)  
- Order of parent classes matters (affects Method Resolution Order - MRO)

```python

# Example for multiple inheritance.
class Father:
    def __init__(self):
        self.father_name = "Mr. Sharma"

    def gardening(self):
        print(f"{self.father_name} is doing gardening")

    def show_father_traits(self):
        print("Father's traits: Hardworking, Patient")


class Mother:
    def __init__(self):
        self.mother_name = "Mrs. Sharma"

    def cooking(self):
        print(f"{self.mother_name} is cooking delicious food")

    def show_mother_traits(self):
        print("Mother's traits: Caring, Creative")
```

```python
class Child(Father, Mother):
  def __init__(self, child_name):
    Father.__init__(self)
    Mother.__init__(self)
    self.child_name = child_name

  def show_talents(self):
    print(f"\n{self.child_name}'s inherited talents:")
    self.gardening()          # From Father
    self.cooking()            # From Mother
    self.show_father_traits() # From Father
    self.show_mother_traits() # From Mother
```

```python
child=Child("Rahul")
```

```python
child.show_father_traits()
```

```python
child.show_mother_traits()
```

```python
child.show_talents()
```
