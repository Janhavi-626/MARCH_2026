```python
## removing leading and trainling whitespaces..
user_input = "    Data Science    "
print(user_input)
cleaned_input=user_input.strip()
print(cleaned_input)
```

```python
## removing only leading whitespace
leading_space="   Janhavi"
print(leading_space)
remove_space=leading_space.lstrip()
print(remove_space)
```

```python
# removing only trailing spaces
trailing_space = "Machine Learning   "
print(trailing_space)
remove_trailing_space=trailing_space.rstrip()
print(f"After trailing:{remove_trailing_space}")
```

```python
# stripping specific chars
course_code = "****DS101************"
new_course_code=course_code.strip("*")
print(new_course_code)
```

```python
## using lstrip with specific char
filename = "__init__.py"
new_filename=filename.rstrip("_")
print(new_filename)
```

```python
## using rstrip with specific char
filenamee = "__init__"
new_filenamee=filenamee.rstrip("_")
print(new_filenamee)
```

```python
## using lstrip with specific char
filenamee = "__init__"
new_filenamee=filenamee.lstrip("_")
print(new_filenamee)
```
