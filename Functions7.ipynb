# Functions (user defined function)

- built-in (comes with python installation, such len(), print(), sum())

A function is a reusable block of code that performs a specific task. Instead of writing the same code multiple
times, you write it once in a function and call it whenever needed.

```python
## Calulate avg of three diff groups..

# group 1

group1_sum = 85 + 90 + 78
group1_avg = group1_sum / 3
print(f"Group 1 Average: {group1_avg}")
```

```python
group2_sum = 78 + 98 +35
group2_avg = group2_sum / 3
print(f"Group 2 Average: {group2_avg}")
```

```python
group3_sum = 12 + 32 + 34
group3_avg = group3_sum / 3
print(f"Group 3 Average: {group3_avg}")
```

```python
## same logic with function
def calculate_average(scores):
  """
  This functions takes list of numbers and returns the average
  """
  total = sum(scores)
  avg = total/len(scores)
  return avg

print(calculate_average([12,23,34]))
print(calculate_average([89,323,54,]))
print(calculate_average([89,34,465]))
print(calculate_average([89,23,45,56,6,34,23,12,]))
```

**def** : keyword to define a function..

**paramters** : Input values (optional)

**docstring** = Documentation string(optional but recomended)

**Function_body** : Code that executes when function is called..

**return** : Sends result back to caller (optional)

## Types of function arguments..


1. Positional Arguments..

Arguments passed in the correct order

```python
def calculate_bmi(weight, height):
  """Calculates Body Mass Index"""
  bmi = weight / (height ** 2)
  return bmi
```

```python
# Positional argument order matters.

bmi1 = calculate_bmi(70, 1.75)    # (kg, meters)
bmi1
```

```python
bmi2 = calculate_bmi(1.75, 70)
bmi2
```

```python
## Keyword Arguments
# Arguments passed with parameter name (order doesn't matter)

def student_info(name, age, grade):
  print(f"Name: {name}")
  print(f"age: {age}")
  print(f"grade: {grade}")
```

```python
student_info(name = "janvi", age = 20, grade = "A")
```

```python
student_info(age = 25, grade = "A", name = "boomi")
```

```python
student_info("pooja",30, "B+")
```

```python
student_info(30, "B+", "Sandhya")
```

## Variable lenght Args(*args)

- Accept any number of positional arguments..

```python
def calculate_sum(*numbers):
  print(type(numbers))
  total = 0
  for num in numbers:
    total += num
  return total
```

```python
numbers = (10,20,30,40,50,60)

total = 0
for num in numbers:
  total += num

print(total)
```

```python
calculate_sum(1,2,3)
```

```python
calculate_sum(12,23,34,45,45,56,6,67,76)
```

```python
calculate_sum(18)
```

## Variable lenght keyword Args(**kwargs)

- Accept any number of keyword arguments.

```python
def student_detail(**details):
  print("Student information: ")
  for key, value in details.items():
    print(f"{key}: {value}")
```

```python
student_detail(name= "Bob", age = 20, grade = "A", city = "Mumbai")
```

```python
## combining all args typs

# order matters: positional, *args, default, **krargs

def complex_function(a,b, *args, c =10, **kwargs):
  print(f"a (position): {a}")
  print(f"b (position): {b}")
  print(f"agrs(variable positional): {args}")
  print(f"c (defaul): {c}")
  print(f"kwargs(vaiable keyword: {kwargs})")
```

```python
complex_function(1,2,3,4,5,c=20, x=100, y=200, z= 300)
```
