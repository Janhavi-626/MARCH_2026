```python
## Calculation of time of all the functions with the help of decorator function
import time
def timer(func):
    def wrapper(*args,**kwargs):
        start=time.time()
        result=func(*args,**kwargs)
        end=time.time()
        print(f"[TIME]took {end-start:.4f}seconds")
        return result
    return wrapper
```

```python
#function without decorator
def process_data_without_decorator():
    total = 0
    for i in range(100000000):
        total += i
    return total
```

```python
process_data_without_decorator()
```

```python
#function with time decorator
@timer
def process_data_with_decorator():
    total=0
    for i in range(100000000):
        total+=i
    return total
```

```python
process_data_with_decorator()
```

#### Syntax for python decorator



```python
def decorator_function(original_function):
    def wrapper_function():
        print("Something before function execution")
        original_function()
        print("Something after function execution")
    return wrapper_function
```

```python
def say_hello():
  print("Hello")
```

```python
say_hello.__name__
```

```python
say_hello()
```

```python
@decorator_function
def say_hello():
  print("Hello")
```

```python
say_hello()
```

```python
@timer
def another_function():
  time.sleep(3)
  print("Task done")
```

```python
another_function.__name__
```

```python
another_function()
```

### Example for logger

```python
def logger(func):
  def wrapper(*args, **kwargs):
    print(f"Function {func.__name__} called with {args}")
    return func(*args, **kwargs)
  return wrapper

# logging is the process of noting down what function did and which function did it
```

```python
def multiply(a, b):
  return a * b

def addition(a,b,c):
  return a + b + c
```

```python
multiply(5,7)
```

```python
addition(4,5,7)
```

```python
@logger
def multiply(a, b):
  return a * b

@logger
def addition(a,b,c):
  return a + b + c
```

```python
multiply(8,9)
```

```python
addition(1,2,3)
```

## Inhertance in OOPs

Inheritance is a mechanism where a new class (child/derived class) can inherit properties and methods from an existing class (parent/base class). This promotes code reusability and establishes a relationship between classes.





## Advantages of Inheritance

### 1. Code Reusability
- Write common code once in the parent class  
- All child classes automatically inherit that code  
- Follows the DRY (Don't Repeat Yourself) principle  

### 2. Extensibility
- Easy to add new features by creating new child classes  
- Existing code remains unchanged  

### 3. Maintainability
- Changes in the parent class reflect in all child classes  
- Fix bugs in one place instead of multiple places  

### 4. Logical Hierarchy
- Represents real-world relationships (IS-A relationship)  
- Makes code more intuitive and structured  

### 5. Polymorphism Support
- Forms the foundation for polymorphism  
- Enables flexible and dynamic behavior in code
