```python
print("Palindrome Checker")

n = int(input("Enter a number: "))

original = n
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10

if original == reverse:
    print("Palindrome Number")
else:
    print("Not a Palindrome Number")
```

```python
print("Armstrong Number Checker")

n = int(input("Enter a number: "))

original = n
sum_digits = 0

# Step 1: find number of digits
temp = n
count = 0

while temp > 0:
    count = count + 1
    temp = temp // 10

# Step 2: calculate sum of digits^power
temp = n

while temp > 0:
    digit = temp % 10
    sum_digits = sum_digits + (digit ** count)
    temp = temp // 10

# Step 3: check result
if sum_digits == original:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
print("Diamond Pattern")

n = int(input("Enter odd number of rows: "))

mid = (n + 1) // 2

# Upper half (including middle row)
for i in range(1, mid + 1):
    for j in range(1, mid - i + 1):
        print(" ", end="")
    for k in range(1, 2 * i):
        print("*", end="")
    print()

# Lower half
for i in range(mid - 1, 0, -1):
    for j in range(1, mid - i + 1):
        print(" ", end="")
    for k in range(1, 2 * i):
        print("*", end="")
    print()
```

```python
print("LCM Calculator")

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

x = a
y = b

# Step 1: Find GCD using Euclidean algorithm
while y != 0:
    remainder = x % y
    x = y
    y = remainder

gcd = x

# Step 2: Find LCM using formula
lcm = (a * b) // gcd

print("LCM is:", lcm)
```

```python
print("Number to Words")

n = int(input("Enter a number (0-999): "))

ones = ["", "One", "Two", "Three", "Four", "Five", "Six", "Seven", "Eight", "Nine"]
teens = ["Ten", "Eleven", "Twelve", "Thirteen", "Fourteen", "Fifteen",
         "Sixteen", "Seventeen", "Eighteen", "Nineteen"]
tens = ["", "", "Twenty", "Thirty", "Forty", "Fifty", "Sixty", "Seventy", "Eighty", "Ninety"]

if n == 0:
    print("Zero")

elif n < 10:
    print(ones[n])

elif n < 20:
    print(teens[n - 10])

elif n < 100:
    t = n // 10
    o = n % 10
    print(tens[t], ones[o])

else:
    h = n // 100
    rem = n % 100

    result = ones[h] + " Hundred"

    if rem != 0:
        if rem < 10:
            result = result + " " + ones[rem]
        elif rem < 20:
            result = result + " " + teens[rem - 10]
        else:
            t = rem // 10
            o = rem % 10
            result = result + " " + tens[t]
            if o != 0:
                result = result + " " + ones[o]

    print(result)
```

```python
print("Strong Number Checker")

n = int(input("Enter a number: "))

original = n
sum_fact = 0

while n > 0:
    digit = n % 10

    # calculate factorial of digit
    fact = 1
    for i in range(1, digit + 1):
        fact = fact * i

    sum_fact = sum_fact + fact
    n = n // 10

if sum_fact == original:
    print("Strong Number")
else:
    print("Not a Strong Number")
```

```python
print("Caesar Cipher Program")

message = input("Enter message: ")
shift = int(input("Enter shift value: "))

print("1. Encrypt")
print("2. Decrypt")

choice = int(input("Enter choice: "))

result = ""

for char in message:

    if char.isalpha():

        # handle uppercase
        if char.isupper():
            start = ord('A')
        else:
            start = ord('a')

        if choice == 1:   # encrypt
            new_char = chr((ord(char) - start + shift) % 26 + start)
        else:             # decrypt
            new_char = chr((ord(char) - start - shift) % 26 + start)

        result = result + new_char

    else:
        result = result + char

print("Result:", result)
```

```python
print("ATM Machine Simulator")

balance = 10000
pin = 1234
attempts = 3
history = []

# PIN verification
while attempts > 0:
    entered_pin = int(input("Enter PIN: "))

    if entered_pin == pin:
        print("PIN correct!\n")
        break
    else:
        attempts -= 1
        print("Wrong PIN. Attempts left:", attempts)

if attempts == 0:
    print("Card blocked!")
    exit()

# ATM Menu
while True:
    print("\n--- ATM MENU ---")
    print("1. Check Balance")
    print("2. Withdraw Money")
    print("3. Deposit Money")
    print("4. Change PIN")
    print("5. Exit")

    choice = int(input("Enter choice: "))

    # Check balance
    if choice == 1:
        print("Balance:", balance)
        history.append("Checked balance")

    # Withdraw
    elif choice == 2:
        amt = int(input("Enter amount to withdraw: "))

        if amt > 0 and amt <= balance:
            balance -= amt
            print("Withdrawal successful")
            history.append(f"Withdrew {amt}")
        else:
            print("Invalid or insufficient balance")

    # Deposit
    elif choice == 3:
        amt = int(input("Enter amount to deposit: "))

        if amt > 0:
            balance += amt
            print("Deposit successful")
            history.append(f"Deposited {amt}")
        else:
            print("Invalid amount")

    # Change PIN
    elif choice == 4:
        old_pin = int(input("Enter old PIN: "))

        if old_pin == pin:
            new_pin = int(input("Enter new PIN: "))
            pin = new_pin
            print("PIN changed successfully")
            history.append("PIN changed")
        else:
            print("Incorrect old PIN")

    # Exit
    elif choice == 5:
        print("Exiting ATM...")
        break

    else:
        print("Invalid choice")

    # Show last 5 transactions
    print("\nLast transactions:")
    for item in history[-5:]:
        print("-", item)
```

```python
print("Automorphic Number")

n = int(input("Enter a number: "))

square = n * n

temp = n
digits = 0

while temp > 0:
    digits += 1
    temp = temp // 10

last_part = square % (10 ** digits)

if last_part == n:
    print("Automorphic Number")
else:
    print("Not an Automorphic Number")
```

```python
print("Amicable Numbers")

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

sum_a = 0
sum_b = 0

for i in range(1, a):
    if a % i == 0:
        sum_a = sum_a + i

for i in range(1, b):
    if b % i == 0:
        sum_b = sum_b + i

if sum_a == b and sum_b == a:
    print("Amicable Numbers")
else:
    print("Not Amicable Numbers")
```

```python
print("Kaprekar Number")

n = int(input("Enter a number: "))

square = n * n
temp = n

digits = 0

while temp > 0:
    digits += 1
    temp = temp // 10

power = 10 ** digits

right = square % power
left = square // power

if left + right == n:
    print("Kaprekar Number")
else:
    print("Not a Kaprekar Number")
```

```python
print("Spiral Matrix")

n = int(input("Enter size n: "))

matrix = [[0] * n for _ in range(n)]

top = 0
bottom = n - 1
left = 0
right = n - 1

num = 1

while num <= n * n:

    # left to right
    for j in range(left, right + 1):
        if num <= n * n:
            matrix[top][j] = num
            num += 1
    top += 1

    # top to bottom
    for i in range(top, bottom + 1):
        if num <= n * n:
            matrix[i][right] = num
            num += 1
    right -= 1

    # right to left
    for j in range(right, left - 1, -1):
        if num <= n * n:
            matrix[bottom][j] = num
            num += 1
    bottom -= 1

    # bottom to top
    for i in range(bottom, top - 1, -1):
        if num <= n * n:
            matrix[i][left] = num
            num += 1
    left += 1

# print matrix
for row in matrix:
    for val in row:
        print(val, end=" ")
    print()
```

```python
print("Twin Primes")

n = int(input("Enter a number: "))

def is_prime(x):
    if x < 2:
        return False
    for i in range(2, x):
        if x % i == 0:
            return False
    return True

for i in range(2, n - 1):
    if is_prime(i) and is_prime(i + 2):
        print(i, i + 2)
```

```python
print("Problem 11.6: Matrix Addition")

rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

matrix1 = []
matrix2 = []

print("Enter elements of Matrix 1:")
for i in range(rows):
    row = []
    for j in range(cols):
        row.append(int(input()))
    matrix1.append(row)

print("Enter elements of Matrix 2:")
for i in range(rows):
    row = []
    for j in range(cols):
        row.append(int(input()))
    matrix2.append(row)

result = []

for i in range(rows):
    row = []
    for j in range(cols):
        row.append(matrix1[i][j] + matrix2[i][j])
    result.append(row)

print("Result Matrix:")
for row in result:
    print(row)
```

```python
print("Sieve of Eratosthenes")

n = int(input("Enter a number: "))

prime = [True] * (n + 1)
prime[0] = False
prime[1] = False

i = 2

while i * i <= n:
    if prime[i]:
        for j in range(i * i, n + 1, i):
            prime[j] = False
    i += 1

print("Prime numbers are:")

for i in range(2, n + 1):
    if prime[i]:
        print(i)
```

```python
print("Problem 11.10: Sudoku Validator")

board = []

print("Enter 9 rows (each row has 9 numbers separated by space):")
for i in range(9):
    row = list(map(int, input().split()))
    board.append(row)

def is_valid_group(group):
    return set(group) == set(range(1, 10))

valid = True

# 1. Check rows
for i in range(9):
    if not is_valid_group(board[i]):
        valid = False

# 2. Check columns
for j in range(9):
    col = []
    for i in range(9):
        col.append(board[i][j])
    if not is_valid_group(col):
        valid = False

# 3. Check 3x3 boxes
for box_row in range(0, 9, 3):
    for box_col in range(0, 9, 3):
        box = []
        for i in range(3):
            for j in range(3):
                box.append(board[box_row + i][box_col + j])
        if not is_valid_group(box):
            valid = False

if valid:
    print("Valid Sudoku")
else:
    print("Invalid Sudoku")
```
