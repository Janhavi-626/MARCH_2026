## 1. What is Range?

`range()` is a built-in function that generates a sequence of numbers.  
It's **NOT a list**, but a special **range object**.

### Key Features:

- **Immutable**: Cannot be changed after creation  
- **Memory Efficient**: Doesn't store all numbers, generates them on demand  
- **Fast**: Very efficient for loops  
- **Lazy**: Only calculates numbers when you need them  
- **Integer Only**: Works only with whole numbers (no decimals)

## Difference Between `==` and `is` in Python

### 1. `==` (Equality Operator)

- Compares **values**
- Checks if two objects have the **same content**
- Used for **value comparison**

#### Example:
```python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)  # True (values are same)
print(a is b)  # False (different objects)

a = [1, 2, 3]
b = a

print(a == b)  # True (values are same)
print(a is b)  # True (same object in memory)

````python
a = [1, 2, 3]
b = a

print(a == b)  # True (values are same)
print(a is b)  # True (same object in memory)
````

a = [1, 2, 3] b = a

print(a == b) # True (values are same) print(a is b) # True (same object in memory)



````python
a = 10
b = 10
c = b

print(a == b)   # compare if both the values are same
print(a is b)   # check if both belong to same memory
print(b is c)
````

````python
r = range(10, 50, 3)

print(f"Start: {r.start}")
print(f"Stop: {r.stop}")
print(f"Step: {r.step}")
print(f"Length: {len(r)}")
print(f"First element: {r[0]}")
print(f"Last element: {r[-1]}")
````
