```python
rows = 5

# Upper pyramid
for i in range(1, rows + 1):
    print(" " * (rows - i) + "*" * (2 * i - 1))

# Lower pyramid
for i in range(rows - 1, 0, -1):
    print(" " * (rows - i) + "*" * (2 * i - 1))
```

```python
print("Simple chatbot (type 'bye' to exit)")

while True:
    user = input("You: ")

    if user.lower() == "hello":
        print("Bot: Hi there!")

    elif user.lower() == "how are you":
        print("Bot: I am fine, thanks!")

    elif user.lower() == "your name":
        print("Bot: I am a simple chatbot.")

    elif user.lower() == "bye":
        print("Bot: Goodbye!")
        break

    else:
        print("Bot: I don't understand that.")
```
