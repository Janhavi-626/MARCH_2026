```python
# 2. try-except Blocks
# block is a method used to catch and handle exceptions, preventing program crashes.

# The try-except Basic Syntax:
# try:
#     # Code that might raise an exception
#     risky_operation()
# except ExceptionType:
#     # Code to handle the exception
#     handle_error()
```

- try:	Place the code inside this block that might cause an error.
- except:	This block executes only if an exception occurs inside try.
- ExceptionType	Replace with the specific error you want to catch (e.g., ValueError, ZeroDivisionError, etc.).
- handle_error()	Your custom function or code to gracefully handle the error.

```python
# Example 1: Basic Exception Handling

try:
    # Taking input from user and converting it to integer
    # This may raise ValueError if user enters something other than a number
    number = int(input("Enter a number: "))

    # Performing division - may cause ZeroDivisionError if number = 0
    result = 10 / number

    # Displaying the result
    print(f"Result: {result}")

# This block runs if user enters 0 (division by zero is not allowed)
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")

# This block runs if user enters a non-numeric input (like text)
except ValueError:
    print("Error: Please enter a valid number!")
```

```python
# Example 2: Catching Any Exception

try:
    # This line will cause ZeroDivisionError because division by zero is not allowed
    x = 10 / 0

# 'Exception as e' catches any type of error and stores error info in variable 'e'
except Exception as e:
    # Displaying the actual error message using formatted string
    print(f"An error occurred: {e}")
    
    
# try: → Put risky code here.

# except Exception as e: → Catches any error and stores the message in e.

# print(f"... {e}") → Shows what error happened (for debugging/clarity).
```

### 3. Handling Multiple Exceptions

- You can handle multiple exceptions in different ways.

```python
# Method 1: Multiple except Blocks

try:
    # A list containing three elements
    data = [1, 2, 3]

    # Taking input for index and converting to integer
    # If user enters non-numeric input => ValueError occurs
    index = int(input("Enter index: "))

    # Accessing the list element using index
    # If index is out of range => IndexError occurs
    print(data[index])

# Triggered when user enters an index that doesn't exist (like 5 or -10)
except IndexError:
    print("Error: Index out of range!")

# Triggered when user enters non-integer input (like 'a', 'hello')
except ValueError:
    print("Error: Invalid input! Please enter an integer.")
# Catches any other unexpected errors
except Exception as e:
    print(f"Unexpected error: {e}")
```

```python
# Method 2: Tuple of Exceptions (Same Handler)

try:
    # Taking input and converting to integer
    # If user enters text or invalid number => ValueError occurs
    value = int(input("Enter a number: "))

    # Dividing 100 by the entered number
    # If value is 0 => ZeroDivisionError occurs
    result = 100 / value

    print(f"Result: {result}")

# This block will run if either ValueError or ZeroDivisionError occurs
# 'as e' stores the error message in variable 'e'
except (ValueError, ZeroDivisionError) as e:
    print("your custom message")
    print(f"Debug info: {e}")


# Feature	                Description
# Multiple                  exceptions	(ValueError, ZeroDivisionError) handled together
# as e	                    Keeps the original error message
# Program won’t crash	    Instead shows a friendly message
```

```python
# Method 3: Getting Exception Details

try:
    # Trying to open a file in read mode
    # Since the file does not exist, this will raise FileNotFoundError
    file = open("nonexistent.txt", "r")

# This block executes if the file is not found
# 'as e' stores the error object so we can print the message
except FileNotFoundError as e:
    # Printing the actual error message returned by Python
    print(f"File error: {e}")

    # Printing the type of error to understand what exception occurred
    print(f"Error type: {type(e).__name__}")
```

### 4. The finally Clause

- The finally block executes regardless of whether an exception occurred or not. It's commonly used for cleanup
operations like closing files or database connections.

```python
# Syntax

#try:
    # Code that may cause an exception
    # If 'operation()' fails, execution jumps to except block
   # operation()

# This block runs only if a specific exception occurs
# Replace 'ExceptionType' with the actual exception you want to handle
#except ExceptionType:
    # Code to handle the exception
    #handle_error()

#finally:
    # This block always runs whether error occurs or not
    # Useful for cleanup tasks (closing files, DB connections, etc.)
    #cleanup()
```

```python
# Block	                    When it runs	                              Purpose
# try	                    Runs normally	                    Contains risky code that may fail
# except	                Runs only when an error happens	    Handles errors gracefully
# finally	                Runs always	                        Cleanup or closing tasks
```

```python
# Example 1: File Handling

# Initializing file variable so it can be accessed in 'finally' block
file = None

try:
    # Trying to open a file in read mode
    # If the file doesn't exist, FileNotFoundError will occur
    file = open("data.txt", "r")

    # Reading file content
    content = file.read()
    print(content)

# Executes when file is not found
except FileNotFoundError:
    print("File not found!")

# 'finally' block always runs regardless of success or exception
finally:
    # Closing file only if it was opened successfully
    if file:
        file.close()
        print("File closed successfully")
```

```python
# Example 2: Database Connection

import sqlite3    # Importing the SQLite library to work with database

connection = None   # Initialize connection so it's accessible in finally block

try:
    # Establishing connection to database (creates file if not exists)
    connection = sqlite3.connect("database.db")

    # Creating a cursor object to execute SQL queries
    cursor = connection.cursor()

    # Executing a query to fetch all rows from 'users' table
    cursor.execute("SELECT * FROM users")

    # Fetching all results from the executed query
    results = cursor.fetchall()
    print(results)

# Handles any database-related errors like connection issues/SQL errors
except sqlite3.Error as e:
    print(f"Database error: {e}")

# Executes always whether query succeeded or exception occurred
finally:
    # Closing DB connection only if it was successfully opened
    if connection:
        connection.close()
        print("Database connection closed")


# Key Point: finally executes even if there's a return statement in the try or except block.
```

### 5. The else Clause

The else block executes only if no exception was raised in the only when the try block succeeds.

```python
# # Syntax

# try:
#     # Code that may raise an exception
#     # If operation() throws an error, control goes to except block
#     operation()

# # This block runs only when the specified exception occurs
# # Replace 'ExceptionType' with specific exception like ValueError, ZeroDivisionError, etc.
# except ExceptionType:
#     # Code to handle the exception
#     handle_error()

# # This block runs only if NO exception occurred inside try block
# else:
#     # Successful execution flow
#     success_operation()

# # This block always runs, no matter what happens above
# # Useful for cleanup tasks like closing files/DB connections
# finally:
#     cleanup()
```

```python
# Example 1: Division Calculator

try:
    # Taking two numeric inputs from the user
    # If user enters non-numeric values => ValueError will occur
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    # Performing division (may cause ZeroDivisionError if num2 = 0)
    result = num1 / num2

# Runs when input is not a valid number (like text or symbols)
except ValueError:
    print("Invalid input! Please enter numbers.")

# Runs when trying to divide by zero
except ZeroDivisionError:
    print("Cannot divide by zero!")

# Executes only if NO exception occurs in try block
else:
    print(f"Result: {result}")
    print("Calculation successful!")

# Always executes whether exception occurs or not
finally:
    print("Calculator Shut down.")
```

```python
# Example 2: File Processing

try:
    # Using 'with open' automatically handles file closing, even if an error occurs
    with open("data.csv", "r") as file:
        data = file.read()   # Reading the entire file

# Executes when file is missing or path is incorrect
except FileNotFoundError:
    print("File not found!")

# Executes only when NO exception occurs in try block
else:
    print("File read successfully")
    print(f"Data length: {len(data)} characters")
    # Process data here (e.g. parsing CSV, printing lines, etc.)

# Always executes whether exception occurs or not
finally:
    print("File operation completed")
```

```python
# Example 1: Input Validation

# Function to validate age
def set_age(age):
    # Check if age is negative → raise custom exception
    if age < 0:
        raise ValueError("Age cannot be negative!")

    # Check if age is too large to be realistic → raise exception
    if age > 150:
        raise ValueError("Age seems unrealistic!")

    # If valid, return age
    return age


try:
    # Passing an invalid age (-5) → this will trigger ValueError
    user_age = set_age(-5)

# Catching the ValueError raised inside the function
except ValueError as e:
    print(f"Error: {e}")
```

```python
# Example 2: Re-raising Exceptions

# Function that performs division
def divide(a, b):
    try:
        return a / b   # Risky operation: can raise ZeroDivisionError
    
    # Triggered when b = 0
    except ZeroDivisionError:
        print("Logging: Division by zero attempted")  # Internal logging
        raise  # Re-raises the same exception to be handled outside


try:
    # Calling divide with 0 → will raise ZeroDivisionError
    result = divide(10, 0)

# Exception caught in outer scope after being re-raised from inside divide()
except ZeroDivisionError:
    print("Handled in outer scope")
```

```python
# Example 3: Raising Different Exception

# Function to process/convert data to integer
def process_data(data):
    try:
        # Trying to convert data to integer
        return int(data)

    except ValueError:
        # If conversion fails, raise a different (more meaningful) exception
        # This is called exception transformation
        raise TypeError(f"Cannot convert '{data}' to integer")


try:
    # Calling the function with invalid data -> will trigger ValueError inside function
    process_data("abc")

except TypeError as e:
    # The TypeError raised inside function is caught here
    print(e)
```

# Basic Custom Exception:


Best Practices
- 1. Be Specific:     Catch specific exceptions rather than using bare except: or catching Exception .
- 2. Don't Silence Exceptions:    Always handle exceptions appropriately; don't use empty except blocks.
- 3. Use finally for Cleanup:     Always use finally for cleanup operations like closing files or connections.
- 4. Log Exceptions:      In production code, log exceptions for debugging purposes.
- 5. Create Custom Exceptions:    For complex applications, create a hierarchy of custom exceptions.
- 6. Don't Catch SystemExit or KeyboardInterrupt:     Let these propagate normally.
- 7. Use else Wisely:     Use else clause to separate success code from error handling.
- 8. Fail Fast:   Raise exceptions as soon as you detect an error condition.
