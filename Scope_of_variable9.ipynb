```python
# Modifying Global Variable

count = 0    # global variable

def increment():
  global count    # Declare that we want to modify global variable
  count += 1
  print("count: ", count)
```

## Lambda Functions(Anonymous Functions)

A lambda function is a small anonymous function defined in one line..



```python
# Syntax = lambda arguments: expression
```

```python
# regular function

def add(a,b):
 return a + b
```

```python
## Lambda function for same
add_lambda=lambda a,b:a+b
```

```python
print(add(5,6))
print(add_lambda(5,40))
```

```python
## Lambda with single Argument:

square = lambda x: x**2

print(square(7))
```

```python
# Lambda with multiple args

area = lambda length , width: length * width
print(area(10,4))
```

```python
## Lambda with condition

# check is even     # lamda arg: expression1 if condition else expression2

is_even = lambda x: "Even" if x % 2 == 0 else "Odd"

is_even(8)
```

```python
is_even(3)
```

```python
## lamda arg: expression1 if condition else expression2
# Positive/Negative
# Pass/Fail (pass marks >= 50)
# Adult/Minor Age >= 18
```

```python
#Positive or Negative
check = lambda x: "Positive" if x >= 0 else "Negative"

print(check(5))    
print(check(-3))  
```

```python
#pass/fail
result = lambda marks: "Pass" if marks >= 50 else "Fail"

print(result(75)) 
print(result(40))  
```

```python
#Check adult or mainor
age_check = lambda age: "Adult" if age >= 18 else "Minor"

print(age_check(20)) 
print(age_check(15))  
```

```python
# Using lambda with built in fucntions
# map()-
# Apply a function to every elemt in a list(or itreable)
# syntax: map(function, iterable) returns: tranfomerd values

# numebrs = [10,20,30,40]

# "Takes each item -> apply logic -> return new list"
```

```python
## without lambda

def square(x):
  return x * x

numbers = [1,2,3,4,5]

result = list(map(square, numbers)) #map() apply function to every item in a collection # list() convert result to list
result
```

```python
# filter()
# Select only elements that satisfy a condition

# syntax: filter(function, iterable)

# returns only matching items

# "check condition -> keep only True value"
```

```python
numbers = [1,2,3,4,5]

result = list(filter(lambda x: x % 2==0, numbers))
result
```

```python
employees = [
    ("John", 50000, 5),   # (name, salary, experience)
    ("Alice", 70000, 3),
    ("Bob", 50000, 7)
]
# Sort by salary, then experience
# Filter high salary but low experience
```

```python
# Question

# Sort the list based on:

# Salary in descending order

# If salary is the same, sort by experience in ascending order

# Question

# Filter the products where:

# Price is less than 60000

# Rating is greater than 4.5

# Using map(): bold text Add a new field called category: rating ≥ 8 → "Excellent" rating ≥ 7 → "Good" else → "Average"
```

```python
## Sort the list based on:

# Salary in descending order
employees = [
    ("John", 50000, 5),
    ("Alice", 70000, 3),
    ("Bob", 50000, 7),
    ("David", 70000, 2)
]
result1=sorted(employees,key=lambda x:x[1],reverse=True)
print(result1)
```

```python
 ## If salary is the same, sort by experience in ascending order
result1=sorted(employees,key=lambda x:x[2],reverse=True)
print(result1)
```

```python
# Filter the products where:

# Price is less than 60000
result1=list(filter(lambda x:x[1]<60000,employees))
print(result1)
```

```python
# Rating is greater than 4.5
result1=(list(filter(lambda x:x[2]>4.5 , employees)))
print(result1)
```

```python
# Using map(): bold text Add a new field called category: rating ≥ 8 → "Excellent" rating ≥ 7 → "Good" else → "Average"
employees = [
    ("John", 50000, 5),
    ("Alice", 70000, 3),
    ("Bob", 50000, 7),
    ("David", 70000, 2)
]
result2 = list(map( lambda x: ( x[0],x[1],x[2], "High" if x[1] >= 70000 else "Medium" if x[1] >= 50000 else  "Low" ), employees))
print(result2)
```
