## Abstraction

Abstraction in OOP is used when you want to define a set of methods that must be implemented by child classes, while hiding the internal implementation details.


Abstraction allows you to define a blueprint (abstract class) with mandatory methods, ensuring that all derived classes implement those methods in their own way, while hiding how those methods actually work.

```python
from abc import ABC,abstractmethod
class Vehicle(ABC):  #Abstract class
 @abstractmethod
 def start(self):
    pass   #we dont have the logic in this 
 @abstractmethod
 def stop(self):
    pass
@abstractmethod
def dummy_methos(self):
  pass
```

```python
#child class
class Bike(Vehicle):
    def start(self):
        print("bike starts with kick")

    def stop(self):
        print("Bike stops using hand brak")
    def dummy_methos(self):
        pass
```

```python
class Car(Vehicle):
    def start(self):
        print("car starts with kick or self start")

    def stop(self):
        print("car stops using hand break")
    def dummy_methos(self):
        pass
    def accelerate(self):
        print("car is accelerating")
```

```python
bike=Bike()
bike.start()
```

```python
car=Car()
car.start()
```

```python
car.accelerate()
```

# Problem: Abstract Class - Shape

## Objective
Create an abstract class `Shape` and implement different shapes using abstraction.

---

## Requirements

### Abstract Class: `Shape`
- Define an abstract method:
  - `area()`

---

### Child Classes

#### 1. Circle
- Formula:
  - `area = π * r²`

#### 2. Rectangle
- Formula:
  - `area = length * breadth`

---

## What You Should Do
- Use `ABC` (Abstract Base Class)
- Use `@abstractmethod` decorator
- Implement `area()` method in child classes
- Create objects and print the area

```python
from abc import ABC, abstractmethod
@abstractmethod
class Shape(ABC):
    def area(self):
        pass

#child class
class Circla(Shape):
    def __init__(self,r):
        self.r=r

    def area(self):
        print(3.14*self.r*self.r)

class Rectangle(Shape):
    def __init__(self,l,b):
        self.l=l
        self.b=b
    def area(self):
        print(self.l*self.b)

r=Rectangle(4,6)
r.area()

c=Circla(5)
c.area()
```
