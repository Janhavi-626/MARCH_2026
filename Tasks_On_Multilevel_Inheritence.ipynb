```python
class Person:
    def __init__(self,name,age,gender,city,phone):
        self.name=name
        self.age=age
        self.gender=gender
        self.city=city
        self.phone=phone
    def display_person(self):
        print("person details\n")
        print("Name",self.name)
        print("Age",self.age)
        print("Gender",self.gender)
        print("City",self.city)
        print("Phone",self.phone)

    def introduce_person(self):
        print(f"My name is {self.name}, my age is {self.age} and I am living in {self.city}")
    
    def update_Phonenumber(self,new_phone):
        self.phone=new_phone
        print("Phone number is updated to",self.phone)
        
```

```python
#class 2
class Student(Person):
    def __init__(self,name,age,gender,city,phone,studentid,department,course,
                 semester,marks,college):
        super().__init__(name,age,gender,city,phone)
        self.studentid=studentid
        self.department=department
        self.course=course
        self.semester=semester
        self.marks=marks
        self.college=college

    def display_student(self):
        print("Student detaila\n")
        print("Student_Id:",self.studentid)
        print("Department:",self.department)
        print("Course:",self.course)
        print("Semester:",self.semester)
        print("College:",self.college)

    def Cauculate_grade(self):
        if self.marks>=90:
            return "A"
        elif self.marks>=70:
            return "B"
        elif self.marks>=50:
            return "C"
        else:
            return "Fail"
        
    def Check_pass_fail(self):
        if self.marks>=50:
            print("Pass")
        else:
            print("Fail")

    def Updatesemester(self):
        self.semester += 1
        print("Semester updated", self.semester)

    def Showmarks(self):
        print("Marks",self.marks)

```

```python
#class 3
class GraduateStudent(Student):
    def __init__(self,name,age,gender,city,phone,studentid,department,course,semester,marks,college,
                researchtopic,guidename,thesistitle,specialization,publications):
        super().__init__(name,age,gender,city,phone,studentid,department,course,semester,marks,college)
        self.researchtopic=researchtopic
        self.guidename=guidename
        self.thesistitle=thesistitle
        self.specialization=specialization
        self.publications=publications
    def display_research(self):
        print("Research Details\n")
        print("Reasearch Topic:",self.researchtopic)
        print("Guide:",self.guidename)
        print("Thesis Title:",self.thesistitle)
        print("Publivations:",self.publications)
    
    def Show_Specialization(self):
        print("Specialization:",self.specialization)

    def Submitting_thesis(self):
        print(f"Thesis'{self.thesistitle}' submitted successfully")

    def Track_research_process(self):
        print(f"Publications Completed: {self.publications}")
```
