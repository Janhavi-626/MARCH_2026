```python
# Question 3: Calculate the compound interest on $1000 at 5.5% for 3 years

# #P = principal (initial money) = 1000
# R = rate = 5.5%
# n = time (years) = 3
# A = final amount
# Compound Interest = A − P

# A=p(1+R/100)power n
# P = principal (initial money)
# R = interest rate
# n = time (years)
P = 1000
R = 5.5
n = 3
A = P * (1 + R/100) ** n
CI = A - P

print(CI)
```

```python
# Question 4: Find the hypotenuse of a right triangle with sides 3.5 and 4.2
#To find the hypotenuse of a right triangle, we use Pythagoras theorem
#Formula
#c = √(a² + b²)
#a = 3.5
#b = 4.2
#c = hypotenuse
a = 3.5
b = 4.2
c = (a**2 + b**2) ** 0.5
print(c)
```

```python
# Question 5: Calculate the volume of a sphere with radius 7.8
# V = (4/3) π r³
#   or
# V = 4πr³ / 3
r = 7.8
pi = 3.1416

V = (4/3) * pi * (r ** 3)

print(V)
```

```python
# Question 9: Calculate the simple interest: Principal=2500, Rate=6.5%, Time=2.5 years
#Formula
# SI = (P × R × T) / 100
#P = Principal = 2500
# R = Rate = 6.5%
# T = Time = 2.5 years
P = 2500
R = 6.5
T = 2.5

SI = (P * R * T) / 100

print(SI)
```
