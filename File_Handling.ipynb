## What is file handling? 

When a Python program runs, all data (variables, lists, etc.) lives in RAM. When the program ends, that data disappears forever. File handling solves this problem — it lets us save data permanently to the disk, and read it back whenever we need it.
Python makes file handling very easy using the built-in open() function. No extra libraries are needed for basic text files.


**File Handling**: The process of performing operations (create, open, read, write, append, close, delete) on files stored on disk.

**File Object**: An object returned by open(). It acts as a connection between your Python program and the file on disk.

**File Path**: The address of the file on disk. Can be absolute (full path: C:/data/file.txt) or relative (just: file.txt, meaning same folder as your script).

## Opening Files — open() and All Modes

The open() function is the gateway to all file operations. You must open a file before reading or writing. The mode you choose tells Python exactly what you plan to do.

open(filename, mode): Built-in Python function that opens a file and returns a file object. The mode controls whether you can read, write, or both.


# basic syntax: 

open(filename, mode)

## 1. Mode 'r' - Read Only 

'r' (Read): Opens a file for reading only. The cursor starts at the beginning. Raises FileNotFoundError if the file does not exist. This is the default mode.

```python
## 'r' - Read Model 
# "r" mode = READ ONLY  

with open("notes.txt", 'r') as f:
    content = f.read()
    print(content)
    print(type(content))
    
    
# What happens step by step:
# 1. Python looks for 'notes.txt' on disk
# 2. If found  → opens it, gives you a file object
# 3. If NOT found → raises FileNotFoundError (program crashes!)
# 4. f.read() reads ALL text from the file into a string
# 5. 'with' block ends → file automatically closed
```

```python
# safe version with error handling:

try:
    with open('notes1.txt', 'r') as f: 
        print(f.read())
except FileNotFoundError: 
    print("File does not exist!")
```

## Mode 'w' — Write (Overwrite)

'w' (Write): Opens a file for writing. If the file does not exist, Python creates it. If it DOES exist, all previous content is ERASED and replaced. Be careful!


## Danger! : Mode 'w' deletes all existing content in the file the moment you open it! 


```python
with open("scores.txt", 'w') as f: 
    f.write('Alice: 95\n')
    f.write('Bob: 87\n')
```

```python
## Now open in 'w' mode again 

with open('scores.txt', 'w') as f: 
    f.write("Charlie: 92\n")
```

## Mode 'a' — Append
'a' (Append): Opens a file for writing but moves the cursor to the END of the existing content. New data is added after old data. Existing content is never erased. Creates the file if it doesn't exist.


```python
## Day 1 - first run 
with open('attendance.txt', 'a') as f: 
    f.write('Day 1: Alice, Bob, Chalie\n')
```

```python
# Day 2- second run 
with open('attendance.txt', 'a') as f: 
    f.write('Day 2: Alice, Diana, Eve\n')
```

```python
## Day 3; third run 

with open('attendance.txt', 'a') as f: 
    f.write("Day 3: Bob, Charlie, Eve\n")
    
    
# The key difference from 'w':
# 'w' → cursor at START → erases old data
# 'a' → cursor at END   → keeps old data
```

## Mode 'x' — Create (Exclusive)

'x' (Create): Creates a new file and opens it for writing. Raises FileExistsError if the file already exists. Use this when you want to make sure you never accidentally overwrite a file.


```python
try: 
    with open('new_report.txt', 'x') as f:
        f.write("This is a brand new file!\n")
    print("File created successfully!")
except FileExistsError: 
    print('Error: File already exists. Not overwriting.')
```

```python
with open('new_report.txt', 'x') as f:
    f.write("This is a brand new file!\n")
print("File created successfully!")
```

## Mode 'r+' — Read and Write

'r+' (Read+Write): Opens an existing file for both reading and writing. The cursor starts at the beginning. The file MUST already exist. Useful when you need to both read content and update it.


```python
## first creat the file. 

with open('counter.txt', 'w') as f: 
    f.write('count: 0')
```

```python
## Now read the current value and update it 

with open('counter.txt', 'r+') as f: 
    content = f.read() 
    print('Brefore:', content)
    
    f.seek(0)       # Go back to start 
    f.write('Count: 1')     # Overwrite with new value
    
```

## Mode 'rb' and 'wb' — Binary Modes

'rb' (Read Binary): Opens a file for reading in binary mode. Returns bytes objects instead of strings. Used for images, audio, video, and other non-text files.

'wb' (Write Binary): Opens a file for writing in binary mode. Writes bytes objects. Used to save binary data like images or downloads.


```python
# Binary Modes 'rb' and 'wb'
# Binary mode — for images, audio, video, etc.
 
# Reading binary data
with open('janavi.jpeg', 'rb') as f:
    data = f.read()
    print(type(data))          # <class 'bytes'>
    print('Size:', len(data), 'bytes')
    print('First 4 bytes:', data[:4])  # e.g. b'\xff\xd8\xff\xe0' (JPEG header)
 
# Copying a binary file
with open('janavi.jpeg', 'rb') as src:
    with open('janavi.jpeg', 'wb') as dst:
        dst.write(src.read())
 
print('Binary file copied!')
 
# KEY RULE: Text modes ('r','w') work with strings
#            Binary modes ('rb','wb') work with bytes
#            NEVER open an image with 'r' — use 'rb'
```

## read() 
— Read Entire File as One String

**f.read()**: Reads the entire file content and returns it as a single string. After calling read(), the cursor is at the end of the file.

**f.read(n)**: Reads and returns only n characters from the current cursor position.


```python
## read() and read(n) 

with open("poem.txt", 'w') as f:
    f.write('Roses are red,\n')
    f.write('Violets are blue,\n')
    f.write('Python is great,\n')
    f.write('And so are you,\n')
```

```python
with open("poem.txt", 'r') as f:
    everything = f.read()
    print(everything)
    print(type(everything))
```

```python
# read(n) 

with open("poem.txt", 'r') as f:
    first6 = f.read(6)
    print(first6)
    
    next5 = f.read(5)
print(next5)
```

## readline() — Read One Line at a Time

**f.readline()**: Reads one line from the file including the \n character at the end. Each call reads the NEXT line. Returns empty string '' when end of file is reached.


```python
# readline() example

with open('poem.txt', 'r') as f:
    line1 = f.readline()
    line2 = f.readline()
    line3 = f.readline()
    line4 = f.readline()
    line5 = f.readline()
    
    print(repr(line1))
    print(repr(line2))
    print(repr(line5))
    
    
# repr() - repr() returns the official string representation of an object
    
```

## Writing to files 

**write()** — Write a String
f.write(string): Writes the given string to the file. IMPORTANT: Does NOT add a newline (\n) automatically — you must add it yourself.


```python
with open('shopping.txt', 'w') as f: 
    f.write("Apple\n")      # \n needed for new line
    f.write("Mango\n\n\n")
    f.write("Bread\n")
    
```

**writelines() **

— Write a List of Strings

**f.writelines(list)**: Writes each element of the list to the file one after another. Does NOT add newlines between elements — you must include \n in each string.


```python
with open('shopping.txt', 'w') as f: 
    f.write("Apple\n")      # \n needed for new line
    f.write("Mango\n\n\n")
    f.write("Bread\n")
    
```

```python
fruits = ['Apple\n', 'Banana\n', 'Cherry\n', 'Date\n'] 

with open('fruits.txt', 'w') as f:
    f.writelines(fruits)
```
