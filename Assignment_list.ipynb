```python
# Question 15: Transpose the matrix [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
m = [[1,2,3],[4,5,6],[7,8,9]]

result = []

for i in range(len(m[0])):   # columns
    temp = []
    for j in range(len(m)):  # rows
        temp.append(m[j][i])
    result.append(temp)

print(result)
```

```python
# Question 34: Count the depth of nesting in [1, [2, 3], [4, [5, 6]], 7]
def depth(lst):
    if not isinstance(lst, list):
        return 0
    
    max_depth = 1
    
    for item in lst:
        max_depth = max(max_depth, 1 + depth(item))
    
    return max_depth

data = [1, [2, 3], [4, [5, 6]], 7]

print(depth(data))
```
